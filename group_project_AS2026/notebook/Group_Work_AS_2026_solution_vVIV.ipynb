# 3. One worked model: forecasting the next twelve months' return (given)

This section shows the pattern you must follow in all your own models.

**The decision-time rule.** At the moment the decision is made, which columns are already known? We form portfolios on **1 July** of each year. By then every firm has published its previous fiscal year (the latest `Publication_Date` is at the end of May). So at formation in July of year $t+1$:

- every **feature** comes from the fiscal-year row of year $t$ (or earlier), plus returns up to June of $t+1$;
- the **target** is the total return from July of $t+1$ to June of $t+2$, compounded from the monthly file;
- the **target** needs twelve holding months to be well defined, so the model is fitted on firms that
  have them; firms that leave mid-year stay in the panel for the backtest in Section 4,
- the **split** is by formation year: train on formation years up to 2021, test on 2022 to 2024. A random split would put a firm's neighbouring years on both sides and let the model recognise firms instead of forecasting.

Every column you add as a feature in task (a) has to pass this rule. Ask of each one: *is it in a file that existed on 1 July, with the value it had then?*

`Year` is not a feature. Industry and region are one-hot encoded so that a linear model can use them (task b). The feature list below is deliberately short; extending it is your work.


```python
FORMATION_MONTH = 7
SCORES = ["ESG_Score", "Environmental_Score", "Social_Score", "Governance_Score"]

def holding_returns(fm, factors, formation_year):
    # 12-month total and excess return from July of formation_year to June of the next year
    start, end = f"{formation_year}-07", f"{formation_year + 1}-06"
    rf = factors.set_index("Month")["Rf_Pct"]
    w = fm[(fm["Month"] >= start) & (fm["Month"] <= end)].copy()
    w["gross"] = 1 + w["Return_Pct"] / 100
    w["gross_ex"] = 1 + (w["Return_Pct"] - w["Month"].map(rf)) / 100
    g = w.groupby("Company_ID")
    return pd.DataFrame({"Hold_Return_Pct": 100 * (g["gross"].prod() - 1),
                         "Hold_Excess_Pct": 100 * (g["gross_ex"].prod() - 1), "Months": g.size()}).reset_index()

def formation_panel(fy, fm, factors, features_from_panel):
    # one row per firm and formation year; features from fiscal year t, target = next 12-month return
    fy = fy.sort_values(["Company_ID", "Year"]).copy()
    fy["Growth_Lag1"] = fy.groupby("Company_ID")["Growth_Rate"].shift(1)
    fy["Formation_Year"] = fy["Year"] + 1
    published = pd.to_datetime(fy["Publication_Date"]) < pd.to_datetime(fy["Formation_Year"].astype(str) + f"-{FORMATION_MONTH:02d}-01")
    fy = fy[published]
    fy["log_Revenue"] = np.log(fy["Revenue"]); fy["log_Market_Cap"] = np.log(fy["Market_Cap"])
    years = sorted(fy["Formation_Year"].unique())
    past = pd.concat([holding_returns(fm, factors, y - 1).assign(Formation_Year=y) for y in years])
    past = past.rename(columns={"Hold_Return_Pct": "Past12_Return_Pct"})[["Company_ID", "Formation_Year", "Past12_Return_Pct"]]
    tgt = pd.concat([holding_returns(fm, factors, y).assign(Formation_Year=y) for y in years])
    # Keep every firm with at least one holding month, including the ones that leave during the year.
    tgt = tgt.rename(columns={"Months": "Hold_Months"})[
        ["Company_ID", "Formation_Year", "Hold_Return_Pct", "Hold_Excess_Pct", "Hold_Months"]]
    out = fy.merge(past, on=["Company_ID", "Formation_Year"], how="left").merge(tgt, on=["Company_ID", "Formation_Year"], how="inner")
    dummies = pd.get_dummies(out[["Industry", "Region"]], drop_first=True, dtype=int)
    out = pd.concat([out, dummies], axis=1)
    features = features_from_panel + ["Growth_Lag1", "log_Revenue", "log_Market_Cap", "Past12_Return_Pct"] + list(dummies.columns)
    return out.dropna(subset=features + ["Hold_Return_Pct"]).reset_index(drop=True), features

ml_data, feature_cols = formation_panel(panel, returns, factors, ["Growth_Rate", "Profit_Margin"] + SCORES)

# A firm needs a full twelve holding months to be a well-defined forecasting target, so we fit on
# those. Firms that leave during the holding year stay in ml_data: Section 4 holds them until their
# last month, which is what a real backtest does. Dropping them here instead would quietly remove the
# firms that did worst before they disappeared.
model_data = ml_data[ml_data["Hold_Months"] == 12]
print(f"Firm-formation-years: {len(ml_data):,} ({len(model_data):,} with a full twelve holding months), "
      f"formation years {ml_data['Formation_Year'].min()}-{ml_data['Formation_Year'].max()}")
print(f"Features ({len(feature_cols)}): {feature_cols}")

train_mask = model_data["Formation_Year"] <= 2021
test_mask = model_data["Formation_Year"] >= 2022
X, y, groups = model_data[feature_cols], model_data["Hold_Return_Pct"], model_data["Company_ID"]
X_train, y_train, g_train = X[train_mask], y[train_mask], groups[train_mask]
X_test, y_test = X[test_mask], y[test_mask]
print(f"Training rows: {len(X_train):,}; test rows: {len(X_test):,}")

baseline_pred = np.full(len(y_test), y_train.mean())
baseline_rmse = np.sqrt(mean_squared_error(y_test, baseline_pred))
print(f"\nNaive baseline (training mean = {y_train.mean():.2f}%): test RMSE = {baseline_rmse:.3f}")

cv = GroupKFold(n_splits=5)
candidates = {
    "RF depth 3": RandomForestRegressor(n_estimators=300, max_depth=3, min_samples_leaf=20, random_state=42, n_jobs=-1),
    "RF depth 6": RandomForestRegressor(n_estimators=300, max_depth=6, min_samples_leaf=20, random_state=42, n_jobs=-1),
}
cv_results = {}
for name, model in candidates.items():
    scores = cross_val_score(model, X_train, y_train, cv=cv, groups=g_train, scoring="neg_root_mean_squared_error")
    cv_results[name] = (-scores.mean(), scores.std())
    print(f"{name:<12} CV RMSE = {-scores.mean():.3f} (sd across folds {scores.std():.3f})")
best_name = min(cv_results, key=lambda k: cv_results[k][0])
rf_model = candidates[best_name].fit(X_train, y_train)
pred_test = rf_model.predict(X_test)
summary = pd.DataFrame({
    "RMSE": [np.sqrt(mean_squared_error(y_train, rf_model.predict(X_train))), cv_results[best_name][0], np.sqrt(mean_squared_error(y_test, pred_test)), baseline_rmse],
    "R2": [r2_score(y_train, rf_model.predict(X_train)), np.nan, r2_score(y_test, pred_test), r2_score(y_test, baseline_pred)],
}, index=["Train (in-sample)", "CV (grouped, train years)", "Test (2022-2024, once)", "Naive baseline on test"])
print(f"\nSelected on CV: {best_name}\n"); print(summary.round(3))

importance = pd.Series(rf_model.feature_importances_, index=feature_cols).sort_values(ascending=False)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
importance.head(10)[::-1].plot(kind="barh", ax=axes[0]); axes[0].set_title("Top 10 feature importances"); axes[0].set_xlabel("Importance")
axes[1].scatter(y_test, pred_test, alpha=0.3, s=10); lims = [y_test.min(), y_test.max()]
axes[1].plot(lims, lims, "r--", label="perfect forecast"); axes[1].set_xlabel("Actual 12-month return (%)"); axes[1].set_ylabel("Forecast (%)")
axes[1].set_title(f"Test formation years 2022-2024 (R2 = {r2_score(y_test, pred_test):.3f})"); axes[1].legend()
plt.tight_layout(); plt.show()
```

**Reading the result.** Returns are hard to forecast: in this run the selected forest's test RMSE is 37.307 percentage points against 36.931 for the naive mean, and the test R2 is -0.480 (your numbers may differ slightly). The forest does not beat the naive mean here, and that is the honest result, not a bug. Two things are worth understanding rather than working around. First, a negative R2 means the model also does worse than simply predicting the test years' own average; that happens whenever the average return in the test period differs from the training period, which is what a market downturn inside the test years does. Second, a model that does much better than this on the test years has usually seen something it should not have. # 4. One backtest pattern: formed after publication, held twelve months (given)

A portfolio rule must use only information available when the portfolio is formed. The pattern below ranks all firms on 1 July of each year on a score from the fiscal-year row published before that date, holds the top decile equally weighted until the next 30 June, and records the monthly portfolio return. Firms that leave the panel during the year contribute their returns until their last month; the remaining names are re-weighted equally. A transaction cost of 20 basis points per unit of one-way turnover is charged at each rebalancing (sell and buy, so 40 basis points on the replaced share).

Over the monthly series we report the annualised return, the annualised volatility, and the **Sharpe ratio** (mean monthly excess return over the risk-free rate divided by its standard deviation, times the square root of twelve). The equal-weight portfolio of all firms in the panel is the benchmark every rule must beat.

The given example ranks on the **ESG score**. Task (d) ranks on your **forecast** from task (b), adds an ESG floor, and asks what the floor costs.


```python
def backtest(picks, fm, factors, cost_bp=20):
    # picks: dict formation_year -> list of Company_ID held from July of that year to June of the next
    rf = factors.set_index("Month")["Rf_Pct"]
    months, rets, turns, prev = [], [], [], set()
    for y in sorted(picks):
        held = set(picks[y])
        turn = 1.0 if not prev else len(held - prev) / max(len(held), 1)
        turns.append(turn)
        window = fm[fm["Company_ID"].isin(held) & (fm["Month"] >= f"{y}-07") & (fm["Month"] <= f"{y + 1}-06")]
        monthly = window.groupby("Month")["Return_Pct"].mean()
        if len(monthly):
            monthly.iloc[0] -= 2 * cost_bp / 100 * turn
        months += list(monthly.index); rets += list(monthly.values); prev = held
    r = pd.Series(rets, index=months)
    ex = r - r.index.map(rf)
    return {"Annualised return (%)": 100 * ((1 + r / 100).prod() ** (12 / len(r)) - 1),
            "Annualised volatility (%)": r.std() * np.sqrt(12),
            "Sharpe ratio": ex.mean() / ex.std() * np.sqrt(12),
            "Mean one-way turnover": np.mean(turns), "Months": len(r)}

formation_years = sorted(y for y in ml_data["Formation_Year"].unique() if y >= 2015)
esg_picks, all_picks = {}, {}
for yr in formation_years:
    at_t = ml_data[ml_data["Formation_Year"] == yr]
    n_top = max(int(0.1 * len(at_t)), 1)
    esg_picks[yr] = at_t.nlargest(n_top, "ESG_Score")["Company_ID"].tolist()
    all_picks[yr] = at_t["Company_ID"].tolist()

results = pd.DataFrame({"Top decile by ESG score": backtest(esg_picks, returns, factors),
                        "All firms, equal weight": backtest(all_picks, returns, factors, cost_bp=0)}).T
print(f"Formation years {formation_years[0]}-{formation_years[-1]}, holding July to June:")
print(results.round(3))
```

**Reading the table.** In this run the ESG-ranked decile earned a Sharpe ratio of 0.175 against 0.212 for the equal-weight panel. Before you read anything into that, remember Section 2: high-ESG firms sit in particular industries and regions, so the decile is partly an industry and region bet; and a Sharpe ratio over 126 monthly observations has a wide confidence band. Task (d) asks you to bootstrap the difference. Note also that ranking the whole world by ESG score is a blunt rule: if a relation between ESG and returns exists only among certain firms, or only in certain years, a single global ranking pooled over the whole sample will bury it. That is what task (c) asks you to look for. And note which firms the backtest was run on: the panel, including the firms that left. Run it on the current universe instead and see what changes.

```python
# ---- (a) Step 1: structure checks on all five files ----
audit = {}   # the numbers collected here feed the audit table and results.json

def rank_corr(a, b):
    # Spearman rank correlation of two aligned series (rows with a missing value are dropped)
    return pd.concat([a, b], axis=1).dropna().corr(method="spearman").iloc[0, 1]

def rank_corr_within_year(df, a, b, year="Year"):
    # rank correlation computed inside each year and averaged over years (removes year-level drift in the scores)
    return float(np.mean([rank_corr(d[a], d[b]) for _, d in df.groupby(year)]))

pub = pd.to_datetime(panel["Publication_Date"])
window_ok = (pub >= pd.to_datetime((panel["Year"] + 1).astype(str) + "-03-31")) & \
            (pub <= pd.to_datetime((panel["Year"] + 1).astype(str) + "-05-31"))
ym = returns.drop_duplicates(["Company_ID", "Month"]).assign(
    m=lambda d: d["Month"].str[:4].astype(int) * 12 + d["Month"].str[5:].astype(int)).sort_values(["Company_ID", "m"])
gap_months = int((ym.groupby("Company_ID")["m"].diff().fillna(1) > 1).sum())

structure = pd.Series({
    "panel: duplicate (Company_ID, Year) keys": int(panel.duplicated(["Company_ID", "Year"]).sum()),
    "restated: duplicate (Company_ID, Year) keys": int(restated.duplicated(["Company_ID", "Year"]).sum()),
    "returns: duplicate (Company_ID, Month) keys": int(returns.duplicated(["Company_ID", "Month"]).sum()),
    "factors: duplicate months": int(factors["Month"].duplicated().sum()),
    "universe: duplicate firms": int(universe["Company_ID"].duplicated().sum()),
    "panel: firms whose Industry or Region changes over time": int(((panel.groupby("Company_ID")["Industry"].nunique() > 1) |
                                                                    (panel.groupby("Company_ID")["Region"].nunique() > 1)).sum()),
    "panel: rows published outside 31 Mar to 31 May of Year+1": int((~window_ok).sum()),
    "returns: months missing inside a firm's history": gap_months,
    "panel: Growth_Rate missing (first year of each firm = firms)": f"{int(panel['Growth_Rate'].isna().sum())} (= {panel['Company_ID'].nunique()} firms)",
})
print(structure.to_string())
print("\nOnly the returns file has duplicated keys (300). The structure checks on the other files are clean; the problems below are in the values.")
```

```python
# ---- (a) Step 2: unit error in Revenue and Market_Cap ----
# Idea: carbon, water and energy scale with revenue within an industry. A firm whose revenue is 1,000 times too large
# looks like it emits 1,000 times too little for its size. The relation is fitted on clearly normal firms (Revenue < 100,000).
ENV = ["Carbon_Emissions", "Water_Usage", "Energy_Consumption"]
lp = panel.assign(l_rev=np.log10(panel["Revenue"]))
resid = pd.DataFrame(index=lp.index)
for c in ENV:
    lp["l_" + c] = np.log10(lp[c])
    normal = lp[(lp["Revenue"] < 1e5) & lp["Carbon_Emissions"].notna()]
    coef = {ind: np.polyfit(d["l_rev"], d["l_" + c], 1) for ind, d in normal.groupby("Industry")}
    slope = lp["Industry"].map({k: v[0] for k, v in coef.items()})
    icept = lp["Industry"].map({k: v[1] for k, v in coef.items()})
    resid[c] = lp["l_" + c] - (slope * lp["l_rev"] + icept)
lp["resid"] = resid.mean(axis=1)                      # missing where the environmental metrics are missing
lp["row_flag"] = lp["resid"] < -1.5                    # about 3 log10 units below the normal relation = factor 1,000
per_firm = lp.groupby("Company_ID").agg(flagged=("row_flag", "sum"), with_env=("resid", "count"))
unit_firms = sorted(per_firm.index[per_firm["flagged"] / per_firm["with_env"].replace(0, np.nan) > 0.5])
unit_mask = panel["Company_ID"].isin(unit_firms)

panel_c = panel.copy()                                 # cleaned copy used from here on
panel_c.loc[unit_mask, ["Revenue", "Market_Cap"]] = panel_c.loc[unit_mask, ["Revenue", "Market_Cap"]] / 1000

top100_before = panel.nlargest(100, "Market_Cap")["Company_ID"].isin(unit_firms).mean()
top100_after = panel_c.nlargest(100, "Market_Cap")["Company_ID"].isin(unit_firms).mean()
audit["unit"] = dict(firms=len(unit_firms), rows=int(unit_mask.sum()),
                     mc_before=panel["Market_Cap"].mean(), mc_after=panel_c["Market_Cap"].mean(),
                     rev_before=panel["Revenue"].mean(), rev_after=panel_c["Revenue"].mean(),
                     top100_before=top100_before, top100_after=top100_after)
print(f"{len(unit_firms)} firms ({int(unit_mask.sum())} firm-years) flagged: {unit_firms}")
print(f"Mean Market_Cap {panel['Market_Cap'].mean():,.0f} -> {panel_c['Market_Cap'].mean():,.0f} USD million; "
      f"mean Revenue {panel['Revenue'].mean():,.0f} -> {panel_c['Revenue'].mean():,.0f}")
print(f"Share of the 100 largest firm-years by market cap that belong to these firms: {top100_before:.0%} -> {top100_after:.0%}")

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
axes[0].hist(lp["resid"].dropna(), bins=80, color="steelblue")
axes[0].set_xlabel("Environmental metrics relative to revenue (log10 units, 0 = typical for the industry)")
axes[0].set_ylabel("Firm-years"); axes[0].set_title("Two clusters: the left one is about 1,000x (3 log10 units) off")
bins = np.linspace(1.4, 8, 100)                       # same bins for both histograms
axes[1].hist(np.log10(panel["Revenue"]), bins=bins, alpha=0.6, label="before the fix", color="grey")
axes[1].hist(np.log10(panel_c["Revenue"]), bins=bins, alpha=0.6, label="after the fix", color="steelblue")
axes[1].set_xlabel("log10 revenue (USD million)"); axes[1].set_ylabel("Firm-years"); axes[1].legend()
axes[1].set_title("Revenue: the second cluster disappears after dividing by 1,000")
plt.tight_layout(); plt.show()
```

```python
# ---- (a) Step 4: Analyst_Outlook carries information from the future ----
assert "panel_c" in globals() and "returns_c" in globals(), "Run the cells above first (Runtime > Run all)."
# Formation panel on the cleaned data, built with the given function (used only for the audit)
ml_c, _ = formation_panel(panel_c, returns_c, factors, base_features + ["Analyst_Outlook"])
rows = []
for label, offset in [("Next 12 months after formation (Jul Y+1 to Jun Y+2): the target", 1),
                      ("12 months before formation (Jul Y to Jun Y+1)", 0),
                      ("The 12 months before that (Jul Y-1 to Jun Y)", -1)]:
    w = pd.concat([holding_returns(returns_c, factors, fy + offset).assign(Year=fy) for fy in range(2012, 2026)])
    w = w[w["Months"] == 12].merge(panel_c[["Company_ID", "Year", "Analyst_Outlook"]], on=["Company_ID", "Year"])
    rows.append((label, rank_corr(w["Analyst_Outlook"], w["Hold_Return_Pct"])))
leak = pd.DataFrame(rows, columns=["12-month window (Y = fiscal year of the row)", "Rank correlation with Analyst_Outlook"])
print(leak.round(3).to_string(index=False))

from sklearn.ensemble import RandomForestRegressor
def rf_test_rmse(features):
    d, ft = formation_panel(panel_c, returns_c, factors, features)
    d = d[d["Hold_Months"] == 12]
    tr, te = d[d["Formation_Year"] <= 2021], d[d["Formation_Year"] >= 2022]
    rf = RandomForestRegressor(n_estimators=300, max_depth=3, min_samples_leaf=20, random_state=42, n_jobs=-1).fit(tr[ft], tr["Hold_Return_Pct"])
    return (np.sqrt(mean_squared_error(te["Hold_Return_Pct"], rf.predict(te[ft]))),
            np.sqrt(mean_squared_error(te["Hold_Return_Pct"], np.full(len(te), tr["Hold_Return_Pct"].mean()))))
rmse_without, rmse_base = rf_test_rmse(base_features)
rmse_with, _ = rf_test_rmse(base_features + ["Analyst_Outlook"])
audit["leak"] = dict(rho_next=leak.iloc[0, 1], rho_before=leak.iloc[1, 1], rho_before2=leak.iloc[2, 1],
                     rmse_without=rmse_without, rmse_with=rmse_with, rmse_base=rmse_base)
print(f"\nRandom forest test RMSE (percentage points): without Analyst_Outlook {rmse_without:.2f}, WITH it {rmse_with:.2f}, naive baseline {rmse_base:.2f}")
print("A column that makes the model look this good, and is related to no other window, behaves as if it was stamped with future information. Excluded.")
```

### Formation panel and features

The decision-time rule is the same as in the worked model: portfolios are formed on 1 July of the formation year; every feature comes from the fiscal-year row published before that date (fiscal year = formation year - 1) or from returns up to June of the formation year; the target is the total return over the following twelve months. The split is by formation year (train up to 2021, test 2022 to 2024).

Six features are new compared with the worked model. Each passes the rule: it is in a file that existed on 1 July, with the value it had then.

```python
# ---- (a) Step 8: our own formation panel (cleaned data, new dated features) ----
assert "panel_c" in globals() and "returns_c" in globals(), "Run the cells above first (Runtime > Run all)."
NEW_FEATURES = ["Env_Change", "Margin_Change", "log_MC_to_Rev", "log_Carbon_Int", "Carbon_Missing", "Past_Vol_Pct"]

def past_window(fm, y):
    # the 12 months before formation: July of y-1 to June of y (returns up to June of the formation year)
    w = fm[(fm["Month"] >= f"{y - 1}-07") & (fm["Month"] <= f"{y}-06")].copy()
    w["gross"] = 1 + w["Return_Pct"] / 100
    g = w.groupby("Company_ID")
    return pd.DataFrame({"Past12_Return_Pct": 100 * (g["gross"].prod() - 1), "Past_Months": g.size(),
                         "Past_Vol_Pct": g["Return_Pct"].std()}).reset_index()

def formation_panel_a(fy, fm, factors):
    fy = fy.sort_values(["Company_ID", "Year"]).copy()
    assert (fy.groupby("Company_ID")["Year"].diff().dropna() == 1).all()      # consecutive years, so shift(1) is the previous year
    g = fy.groupby("Company_ID")
    fy["Growth_Lag1"] = g["Growth_Rate"].shift(1)
    fy["Env_Change"] = fy["Environmental_Score"] - g["Environmental_Score"].shift(1)     # ESG momentum
    fy["Margin_Change"] = fy["Profit_Margin"] - g["Profit_Margin"].shift(1)
    fy["log_Revenue"], fy["log_Market_Cap"] = np.log(fy["Revenue"]), np.log(fy["Market_Cap"])
    fy["log_MC_to_Rev"] = np.log(fy["Market_Cap"] / fy["Revenue"])                       # valuation relative to sales
    fy["log_Carbon_Int"] = np.log(fy["Carbon_Emissions"] / fy["Revenue"])                # carbon per USD million revenue
    fy["Carbon_Missing"] = fy["Carbon_Emissions"].isna().astype(int)
    # gap fill: median of the same fiscal year and industry (all rows of fiscal year t are published before 1 July t+1)
    fy["log_Carbon_Int"] = fy["log_Carbon_Int"].fillna(fy.groupby(["Year", "Industry"])["log_Carbon_Int"].transform("median"))
    fy["Formation_Year"] = fy["Year"] + 1
    pub_ok = pd.to_datetime(fy["Publication_Date"]) < pd.to_datetime(fy["Formation_Year"].astype(str) + f"-{FORMATION_MONTH:02d}-01")
    assert pub_ok.all(), "a row would be used before it was published"
    years = sorted(fy["Formation_Year"].unique())
    past = pd.concat([past_window(fm, y).assign(Formation_Year=y) for y in years])
    past = past[past["Past_Months"] == 12]                                               # require a full 12 months of history
    tgt = pd.concat([holding_returns(fm, factors, y).assign(Formation_Year=y) for y in years]).rename(columns={"Months": "Hold_Months"})
    out = (fy.merge(past.drop(columns="Past_Months"), on=["Company_ID", "Formation_Year"], how="left")
             .merge(tgt[["Company_ID", "Formation_Year", "Hold_Return_Pct", "Hold_Excess_Pct", "Hold_Months"]], on=["Company_ID", "Formation_Year"], how="inner"))
    dummies = pd.get_dummies(out[["Industry", "Region"]], drop_first=True, dtype=int)
    out = pd.concat([out, dummies], axis=1)
    features = (["Growth_Rate", "Profit_Margin"] + SCORES + ["Growth_Lag1", "log_Revenue", "log_Market_Cap", "Past12_Return_Pct"]
                + NEW_FEATURES + list(dummies.columns))
    return out.dropna(subset=features + ["Hold_Return_Pct"]).reset_index(drop=True), features

ml_a, feature_cols_a = formation_panel_a(panel_c, returns_c, factors)
model_data_a = ml_a[ml_a["Hold_Months"] == 12]            # a full twelve holding months: the rows models are fitted and tested on
train_a, test_a = model_data_a["Formation_Year"] <= 2021, model_data_a["Formation_Year"] >= 2022

# decision-time checks that must hold for every row
assert (ml_a["Formation_Year"] == ml_a["Year"] + 1).all()
assert (pd.to_datetime(ml_a["Publication_Date"]) < pd.to_datetime(ml_a["Formation_Year"].astype(str) + "-07-01")).all()
assert "Analyst_Outlook" not in feature_cols_a

print(f"Formation rows: {len(ml_a):,} ({len(model_data_a):,} with a full twelve holding months), formation years {ml_a['Formation_Year'].min()}-{ml_a['Formation_Year'].max()}")
print(f"Features ({len(feature_cols_a)}): {feature_cols_a}")
print(f"Training rows (formation years <= 2021): {int(train_a.sum()):,}; test rows (2022-2024): {int(test_a.sum()):,}")
print(f"Naive baseline: training mean {model_data_a.loc[train_a, 'Hold_Return_Pct'].mean():.2f}%, test RMSE "
      f"{np.sqrt(mean_squared_error(model_data_a.loc[test_a, 'Hold_Return_Pct'], np.full(int(test_a.sum()), model_data_a.loc[train_a, 'Hold_Return_Pct'].mean()))):.3f}")
print(f"Mean 12-month return: training years {model_data_a.loc[train_a, 'Hold_Return_Pct'].mean():.1f}%, test years {model_data_a.loc[test_a, 'Hold_Return_Pct'].mean():.1f}% "
      "(the test years are strong years, not a downturn)")
```

## Task (b): linear baseline vs tree ensembles

**Design.** Target: the total 12-month return (percent) from July of the formation year to June of the next. Features: the 30 features of our formation panel (task a). Split by formation year: training 2015 to 2021, test 2022 to 2024. Models: the naive mean forecast, a linear regression on standardised features, a random forest and gradient boosting. The two ensembles are tuned by grouped cross-validation (5 folds grouped by company, with a fixed fold assignment so that the result does not depend on the scikit-learn version) inside the training years only, over a small grid (random forest: 150 trees, depth 3 or 6, minimum leaf 20 or 60; gradient boosting: 100 trees, learning rate 0.03 or 0.1, depth 2 or 3). The tuning step takes a few minutes on Colab and prints one progress line per setting. The model is **selected on cross-validation**. The test years are evaluated in a single cell and never used to choose anything.

```python
# ---- (b) Step 1: data, cross-validation and tuning of the ensembles ----
assert "model_data_a" in globals(), "Run the cells above first (Runtime > Run all)."
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import ParameterGrid, cross_validate
from scipy.stats import spearmanr

TARGET = "Hold_Return_Pct"          # total 12-month return in percent, July to June after formation
X_tr, y_tr, g_tr = model_data_a.loc[train_a, feature_cols_a], model_data_a.loc[train_a, TARGET], model_data_a.loc[train_a, "Company_ID"]
X_te, y_te = model_data_a.loc[test_a, feature_cols_a], model_data_a.loc[test_a, TARGET]
# Grouped 5-fold cross-validation by company with an explicit, deterministic fold assignment: all rows of one company stay in
# the same fold, and the folds are identical on every machine and scikit-learn version (GroupKFold's own assignment changed
# between versions, which flipped the near-tie between models in a first Colab run).
companies = np.sort(g_tr.unique())
fold_of_company = pd.Series(np.arange(len(companies)) % 5, index=companies)
fold_of_row = g_tr.map(fold_of_company).to_numpy()
cv = [(np.where(fold_of_row != k)[0], np.where(fold_of_row == k)[0]) for k in range(5)]
print("Rows per cross-validation fold:", [len(te) for _, te in cv])

def cv_scores(model):
    res = cross_validate(model, X_tr, y_tr, cv=cv,
                         scoring={"rmse": "neg_root_mean_squared_error", "r2": "r2"})
    return dict(cv_rmse=-res["test_rmse"].mean(), cv_sd=res["test_rmse"].std(), cv_r2=res["test_r2"].mean())

candidates = {"Naive mean (baseline)": DummyRegressor(strategy="mean"),
              "Linear regression": make_pipeline(StandardScaler(), LinearRegression())}
cv_table = {name: cv_scores(m) for name, m in candidates.items()}

grids = {
    "Random forest": (lambda **p: RandomForestRegressor(n_estimators=150, random_state=42, n_jobs=-1, **p),
                      {"max_depth": [3, 6], "min_samples_leaf": [20, 60]}),
    "Gradient boosting": (lambda **p: GradientBoostingRegressor(n_estimators=100, subsample=0.8, min_samples_leaf=20, random_state=42, **p),
                          {"learning_rate": [0.03, 0.1], "max_depth": [2, 3]}),
}
import time
t0 = time.time()
print("Tuning the ensembles with grouped cross-validation. This takes a few minutes on Colab; one line is printed per setting.")
tuning_rows = []
for name, (make, grid) in grids.items():
    for params in ParameterGrid(grid):
        scores = cv_scores(make(**params))
        tuning_rows.append({"model": name, **params, **scores})
        print(f"  {name} {params}: CV RMSE {scores['cv_rmse']:.3f}  ({time.time() - t0:.0f} s elapsed)")
tuning = pd.DataFrame(tuning_rows)

best_params = {}
for name, (make, grid) in grids.items():
    t = tuning[tuning["model"] == name]
    best = t.loc[t["cv_rmse"].idxmin()]
    best_params[name] = {k: (float(best[k]) if k == "learning_rate" else int(best[k])) for k in grid}
    label = f"{name} (tuned)"
    candidates[label] = make(**best_params[name])
    cv_table[label] = dict(cv_rmse=best["cv_rmse"], cv_sd=best["cv_sd"], cv_r2=best["cv_r2"])

print("Cross-validation results of the tuning grid (RMSE in percentage points, grouped 5-fold inside the training years):")
for name in grids:
    t = tuning[tuning["model"] == name].dropna(axis=1, how="all").drop(columns="model")
    print(f"\n{name}:"); display(t.sort_values("cv_rmse").round(3).reset_index(drop=True))
print("Chosen settings:", best_params)
```

```python
# ---- (b) Step 2: train, CV and test side by side (the test years are evaluated here, once) ----
assert "candidates" in globals(), "Run the cells above first (Runtime > Run all)."
test_years = model_data_a.loc[test_a, ["Formation_Year", TARGET]]
rows, fitted = [], {}
for name, m in candidates.items():
    m.fit(X_tr, y_tr)
    fitted[name] = m
    p_tr, p_te = m.predict(X_tr), pd.Series(m.predict(X_te), index=X_te.index)
    # rank correlation between forecast and outcome within each test year (what matters for ranking firms in task d)
    ic = np.nan if np.ptp(p_te.values) < 1e-9 else np.mean([spearmanr(p_te[d.index], d[TARGET])[0] for _, d in test_years.groupby("Formation_Year")])
    rows.append({"Model": name,
                 "Train RMSE": np.sqrt(mean_squared_error(y_tr, p_tr)), "CV RMSE": cv_table[name]["cv_rmse"], "CV sd": cv_table[name]["cv_sd"],
                 "Test RMSE": np.sqrt(mean_squared_error(y_te, p_te)),
                 "Train R2": r2_score(y_tr, p_tr), "CV R2": cv_table[name]["cv_r2"], "Test R2": r2_score(y_te, p_te),
                 "Test rank IC": ic})
comparison = pd.DataFrame(rows).set_index("Model")

selected = comparison.drop(index="Naive mean (baseline)")["CV RMSE"].idxmin()      # selection on CV only
ensembles = [i for i in comparison.index if "tuned" in i]
best_ens = comparison.loc[ensembles, "CV RMSE"].idxmin()
display(comparison.round(3))
print(f"Selected on cross-validation: {selected}.  Best ensemble on cross-validation: {best_ens}.")
y_year_mean = test_years.groupby("Formation_Year")[TARGET].transform("mean")
print(f"For reference: a naive forecast that already knew each test year's mean would have a test RMSE of {np.sqrt(((y_te - y_year_mean) ** 2).mean()):.2f} "
      f"(the plain naive baseline has {comparison.loc['Naive mean (baseline)', 'Test RMSE']:.2f}); the rest of the baseline's error is the shift between training and test years.")
```

## Task (c): where does ESG pay?

**Design.** For every firm and formation year we relate the **previous fiscal year's Environmental score** (standardised within each fiscal year across all firms: mean 0, standard deviation 1) to the **next twelve months' excess return** (total return minus the risk-free rate, July to June, percent). The coefficient is the change in 12-month excess return, in percentage points, for a firm that is one standard deviation higher on the Environmental score than its same-industry peers of that year.

*Sample.* Formation years 2013 to 2024, firms with a full twelve holding months. The Environmental score needs no past-return history, so the sample starts two years earlier than the model panel of tasks (b) and (d). Formation year 2025 has only six months of returns and is left out.

*Estimation.* One regression per region and period (before 2021 and from 2021, defined by formation year), with **industry-by-formation-year fixed effects** (every firm is compared with firms of the same industry in the same year) and standard errors **clustered by company** (one firm's years are not independent evidence). With 14 region-period cells, one "significant" cell can appear by chance, so we also report **Holm-adjusted** p-values across the 14 cells. We repeat everything with the restated scores.

```python
# ---- (c) Step 1: estimation sample ----
assert "panel_c" in globals() and "returns_c" in globals(), "Run the cells above first (Runtime > Run all)."
import statsmodels.formula.api as smf
from statsmodels.stats.multitest import multipletests

def z_within_year(values, years):
    # standardise within each fiscal year across all firms with a row that year
    g = values.groupby(years)
    return (values - g.transform("mean")) / g.transform("std")

sc = panel_c[["Company_ID", "Year", "Region", "Industry", "Environmental_Score"]].merge(
    restated[["Company_ID", "Year", "Environmental_Score"]].rename(columns={"Environmental_Score": "Env_restated"}), on=["Company_ID", "Year"])
sc["z_pit"] = z_within_year(sc["Environmental_Score"], sc["Year"])      # point-in-time score, as first published
sc["z_rest"] = z_within_year(sc["Env_restated"], sc["Year"])            # restated score, as a download today shows it

# next 12-month excess return: fiscal year Y -> formation July Y+1 -> held to June Y+2 (full 12 months only)
tgt = pd.concat([holding_returns(returns_c, factors, f).assign(Year=f - 1) for f in range(2013, 2025)])
tgt = tgt[tgt["Months"] == 12][["Company_ID", "Year", "Hold_Excess_Pct"]]
est = sc.merge(tgt, on=["Company_ID", "Year"]).assign(Formation_Year=lambda d: d["Year"] + 1)
est["Period"] = np.where(est["Formation_Year"] < 2021, "before 2021", "from 2021")
est["ind_year"] = est["Industry"] + "_" + est["Formation_Year"].astype(str)
REGIONS = sorted(est["Region"].unique()); PERIODS = ["before 2021", "from 2021"]

print(f"Estimation sample: {len(est):,} firm-years, {est['Company_ID'].nunique():,} firms, formation years {est['Formation_Year'].min()} to {est['Formation_Year'].max()}")
print("Firm-years by region and period:")
print(est.pivot_table(index="Region", columns="Period", values="Company_ID", aggfunc="size")[PERIODS].to_string())
print(f"Mean standardised score: point-in-time {est['z_pit'].mean():.3f}, restated {est['z_rest'].mean():.3f}; mean 12-month excess return {est['Hold_Excess_Pct'].mean():.2f}%")
```

```python
# ---- (c) Step 2: one regression per region and period, point-in-time and restated scores ----
assert "est" in globals(), "Run the cells above first (Runtime > Run all)."

def one_cell(d, zcol):
    # excess return on the standardised score, with industry-by-year fixed effects, standard errors clustered by company
    m = smf.ols(f"Hold_Excess_Pct ~ {zcol} + C(ind_year)", data=d).fit(
        cov_type="cluster", cov_kwds={"groups": pd.factorize(d["Company_ID"])[0]})
    lo, hi = m.conf_int().loc[zcol]
    return dict(coef=m.params[zcol], se=m.bse[zcol], p=m.pvalues[zcol], lo=lo, hi=hi, n=int(m.nobs), firms=int(d["Company_ID"].nunique()))

rows = []
for version, zcol in [("point-in-time", "z_pit"), ("restated", "z_rest")]:
    for period in PERIODS:
        for region in REGIONS + ["All regions"]:
            d = est[(est["Period"] == period) & ((est["Region"] == region) | (region == "All regions"))]
            rows.append(dict(version=version, region=region, period=period, **one_cell(d, zcol)))
est_table = pd.DataFrame(rows)

# Holm adjustment across the 14 region-period cells, separately for each version (the 'All regions' rows are reference only)
est_table["p_holm"] = np.nan
for version in ["point-in-time", "restated"]:
    mask = (est_table["version"] == version) & (est_table["region"] != "All regions")
    est_table.loc[mask, "p_holm"] = multipletests(est_table.loc[mask, "p"], method="holm")[1]

def fmt(r):
    star = " **" if r["p_holm"] < 0.05 else (" *" if r["p"] < 0.05 else "")
    return f"{r['coef']:+.2f} ({r['se']:.2f}){star}"
wide = {}
for version in ["point-in-time", "restated"]:
    t = est_table[est_table["version"] == version].copy()
    t["cell"] = t.apply(fmt, axis=1)
    wide[version] = t.pivot(index="region", columns="period", values="cell")[PERIODS]
summary_c = pd.concat(wide, axis=1).reindex(REGIONS + ["All regions"])
summary_c.columns = [f"{v}, {p}" for v, p in summary_c.columns]
print("Coefficient in percentage points of 12-month excess return per 1 standard deviation of the Environmental score (standard error in brackets).")
print("* = p < 0.05 for that cell;  ** = still p < 0.05 after the Holm correction across the 14 region-period cells.\n")
display(summary_c)

det = est_table[(est_table["region"] != "All regions")].copy()
print("\nDetail for the point-in-time scores (95% interval, p-values; n = firm-years):")
display(det[det["version"] == "point-in-time"][["region", "period", "coef", "se", "lo", "hi", "p", "p_holm", "n", "firms"]].round(3).reset_index(drop=True))
```

```python
# ---- (c) Step 3: figure ----
assert "est_table" in globals(), "Run the cells above first (Runtime > Run all)."
fig, axes = plt.subplots(1, 2, figsize=(15, 6), sharex=True, sharey=True)
ypos = np.arange(len(REGIONS))[::-1]
for ax, period in zip(axes, PERIODS):
    for k, (version, colour) in enumerate([("point-in-time", "steelblue"), ("restated", "darkorange")]):
        t = est_table[(est_table["version"] == version) & (est_table["period"] == period)].set_index("region").reindex(REGIONS)
        off = (-0.15 if k == 0 else 0.15)
        ax.errorbar(t["coef"], ypos + off, xerr=[t["coef"] - t["lo"], t["hi"] - t["coef"]], fmt="o", color=colour, capsize=3, label=version)
    ax.axvline(0, color="black", lw=0.8)
    ax.set_yticks(ypos); ax.set_yticklabels(REGIONS)
    ax.set_title(f"Formation years {period}")
    ax.set_xlabel("Percentage points of 12-month excess return per 1 standard deviation of the Environmental score\n(95% interval, standard errors clustered by company)")
axes[0].set_ylabel("Region"); axes[0].legend(title="Score used")
fig.suptitle("Intervals are not adjusted for the 14 cells; only Europe from 2021 stays significant after the Holm correction (point-in-time scores)", fontsize=10)
plt.tight_layout(); plt.show()
```

```python
# ---- (c) Step 4: why do the restated scores give different estimates? ----
assert "est" in globals(), "Run the cells above first (Runtime > Run all)."
est["dz"] = est["z_rest"] - est["z_pit"]            # what the restatement added to the standardised score
est["dz_std"] = est["dz"] / est["dz"].std()          # the same, in standard deviations of the restatement itself
print(f"Rank correlation (within each year, averaged over years) with the next 12-month excess return: point-in-time score {rank_corr_within_year(est, 'z_pit', 'Hold_Excess_Pct'):.3f}, "
      f"restated score {rank_corr_within_year(est, 'z_rest', 'Hold_Excess_Pct'):.3f}, the restatement itself (restated minus point-in-time) {rank_corr_within_year(est, 'dz', 'Hold_Excess_Pct'):.3f}")
both = smf.ols("Hold_Excess_Pct ~ z_pit + dz_std + C(ind_year)", data=est).fit(cov_type="cluster", cov_kwds={"groups": pd.factorize(est["Company_ID"])[0]})
print(f"Joint regression with industry-by-year effects (all regions, all years): point-in-time score {both.params['z_pit']:+.2f} (se {both.bse['z_pit']:.2f}), "
      f"restatement part (per 1 standard deviation of the restatement) {both.params['dz_std']:+.2f} (se {both.bse['dz_std']:.2f})")
print(f"The restatement changed {((est['Env_restated'] - est['Environmental_Score']).abs() > 0.01).mean():.0%} of the scores, and what it added is itself related to the return that followed: the restated score behaves as if it contains hindsight (inference from this correlation).")
diff = (est_table[est_table["version"] == "restated"].set_index(["region", "period"])["coef"]
        - est_table[est_table["version"] == "point-in-time"].set_index(["region", "period"])["coef"]).unstack()[PERIODS]
print("\nRestated minus point-in-time coefficient, by region and period (percentage points):")
display(diff.round(2))
```

```python
# ---- (c) Step 5: the one cell that stands out, in more detail (exploratory: chosen after looking at the 14 cells) ----
assert "est" in globals(), "Run the cells above first (Runtime > Run all)."
eu = est[est["Region"] == "Europe"].copy()
eu["from21"] = (eu["Period"] == "from 2021").astype(int)
inter = smf.ols("Hold_Excess_Pct ~ z_pit + z_pit:from21 + C(ind_year)", data=eu).fit(cov_type="cluster", cov_kwds={"groups": pd.factorize(eu["Company_ID"])[0]})
print(f"Europe: is the relation from 2021 different from before 2021? Extra coefficient from 2021 {inter.params['z_pit:from21']:+.2f} (se {inter.bse['z_pit:from21']:.2f}, p = {inter.pvalues['z_pit:from21']:.3f})")

by_year = []
for fy, d in eu.groupby("Formation_Year"):
    m = smf.ols("Hold_Excess_Pct ~ z_pit + C(Industry)", data=d).fit(cov_type="HC1")        # one year: a firm appears once, so robust errors
    lo, hi = m.conf_int().loc["z_pit"]
    by_year.append(dict(formation_year=fy, coef=m.params["z_pit"], se=m.bse["z_pit"], lo=lo, hi=hi, n=int(m.nobs)))
by_year = pd.DataFrame(by_year)
display(by_year.round(2).set_index("formation_year"))
from scipy import stats
late = by_year[by_year["formation_year"] >= 2021]
fm_t = late["coef"].mean() / (late["coef"].std() / np.sqrt(len(late)))
fm_p = 2 * stats.t.sf(abs(fm_t), df=len(late) - 1)
print(f"Europe, from 2021, with years as the unit (Fama-MacBeth over the {len(late)} yearly slopes): mean {late['coef'].mean():+.2f}, t = {fm_t:.2f}, p = {fm_p:.3f} with {len(late) - 1} degrees of freedom. "
      "The firm-clustered p-value of the main table treats firms, not years, as independent.")
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.errorbar(by_year["formation_year"], by_year["coef"], yerr=[by_year["coef"] - by_year["lo"], by_year["hi"] - by_year["coef"]], fmt="o", color="steelblue", capsize=3)
ax.axhline(0, color="black", lw=0.8); ax.axvline(2020.5, color="grey", ls="--", lw=1)
ax.set_xlabel("Formation year (dashed line: start of the 'from 2021' period)")
ax.set_ylabel("Percentage points of 12-month excess return\nper 1 standard deviation of the Environmental score")
ax.set_title("Europe: the estimate year by year (95% interval, industry effects)")
plt.tight_layout(); plt.show()
```

## Task (d): backtest with costs and an ESG floor

**Design.**
- *Rule.* Each year on 1 July, rank the firms of the pool on the forecast of the model selected in task (b), hold the top decile equally weighted until 30 June, and charge 20 basis points per unit of one-way turnover (the starter's convention: the replaced share is sold and bought, 40 basis points on that share, charged in the first month of the holding year).
- *Walk-forward forecasts.* The forecast for formation year Y uses only formation years up to Y-1 (their holding periods end by 30 June of Y, so their outcomes are known on 1 July of Y); the model and its scaler are refitted every year. Formation years 2018 to 2024 (at least three earlier years to train on), 84 months. Formation year 2025 has only six months of returns and is left out.
- *Model.* The pre-set rule picks **linear regression** (lowest CV RMSE, a near-tie). **Gradient boosting** (settings tuned once on the training years of task b: a simplification, because the setting was chosen with years the early forecasts should not have seen, although no return value enters the tuning) is reported as a labelled robustness check, and so is a **linear variant** whose past 12-month return is standardised within each formation year (the pooled version carries year-level variation learned from only a few years).
- *ESG floor.* The pool is restricted to firms whose Environmental score is at or above the 50th, then the 80th percentile of that formation year (computed on the whole pool of that year). The top decile is 10% of the restricted pool, so the 80th-percentile floor holds only about 18 names, which makes it noisy.
- *Universe.* The full panel including the firms that left: forecasts are made for every firm with features (a later exit is unknown on 1 July), and a firm that leaves contributes its returns until its last month, the remaining names being re-weighted equally.
- *Benchmark.* Equal weight on all firms of the formation panel, no cost (as in the starter).
- *Uncertainty.* The Sharpe difference between the best rule and the benchmark is bootstrapped by resampling months (5,000 draws, fixed seed, the same months drawn for the rule and the benchmark).

```python
# ---- (d) Step 1: walk-forward forecasts for formation years 2018 to 2024 ----
assert all(k in globals() for k in ["ml_a", "model_data_a", "best_params", "feature_cols_a", "TARGET", "selected"]), "Run the cells above first (Runtime > Run all)."
BT_YEARS = list(range(2018, 2025))
ml_a["Past12_z"] = ml_a.groupby("Formation_Year")["Past12_Return_Pct"].transform(lambda s: (s - s.mean()) / s.std())
feats_var = ["Past12_z" if f == "Past12_Return_Pct" else f for f in feature_cols_a]
gb_params = dict(n_estimators=100, subsample=0.8, min_samples_leaf=20, random_state=42, **best_params["Gradient boosting"])
MODELS = {
    "Linear regression (selected)": (lambda: make_pipeline(StandardScaler(), LinearRegression()), feature_cols_a),
    "Gradient boosting (robustness check)": (lambda: GradientBoostingRegressor(**gb_params), feature_cols_a),
    "Linear, past return standardised within year (variant)": (lambda: make_pipeline(StandardScaler(), LinearRegression()), feats_var),
}
print("Model selected on cross-validation in task (b):", selected, "(the pre-set rule); the other two are labelled robustness checks.")

bt = ml_a[ml_a["Formation_Year"].isin(BT_YEARS)].copy()          # all firms with features, including firms that later leave
for name, (make, feats) in MODELS.items():
    bt[name] = np.nan
    for y in BT_YEARS:
        train = ml_a[(ml_a["Hold_Months"] == 12) & (ml_a["Formation_Year"] <= y - 1)]     # earlier formation years only
        assert train["Formation_Year"].max() <= y - 1 and len(train) > 0
        model = make().fit(train[feats], train[TARGET])
        rows_y = bt["Formation_Year"] == y
        bt.loc[rows_y, name] = model.predict(bt.loc[rows_y, feats])
print(f"Forecasts for {len(bt):,} firm-years, formation years {BT_YEARS[0]} to {BT_YEARS[-1]}; training years grow from {BT_YEARS[0] - ml_a['Formation_Year'].min()} to {BT_YEARS[-1] - ml_a['Formation_Year'].min()}.")

full12 = bt[bt["Hold_Months"] == 12]
ic = pd.DataFrame({name: {y: rank_corr(d[name], d[TARGET]) for y, d in full12.groupby("Formation_Year")} for name in MODELS})
ic.loc["average 2018-2021"] = ic.loc[2018:2021].mean(); ic.loc["average 2022-2024"] = ic.loc[2022:2024].mean(); ic.loc["average 2018-2024"] = ic.loc[2018:2024].mean()
print("Rank correlation between forecast and the realised 12-month return, by formation year (firms with a full 12 months):")
display(ic.round(3))
```

```python
# ---- (d) Step 2: portfolios, costs and the results table ----
assert "bt" in globals(), "Run the cells above first (Runtime > Run all)."
rf_series = factors.set_index("Month")["Rf_Pct"]

def run_rule(picks, fm, cost_bp=20):
    # same mechanics as the given backtest(), but it returns the monthly return series (needed for the figure and the bootstrap)
    months, rets, turns, prev = [], [], [], set()
    for y in sorted(picks):
        held = set(picks[y])
        turn = 1.0 if not prev else len(held - prev) / max(len(held), 1)
        window = fm[fm["Company_ID"].isin(held) & (fm["Month"] >= f"{y}-07") & (fm["Month"] <= f"{y + 1}-06")]
        monthly = window.groupby("Month")["Return_Pct"].mean().copy()
        monthly.iloc[0] -= 2 * cost_bp / 100 * turn                 # sell and buy the replaced share
        months += list(monthly.index); rets += list(monthly.values); turns.append(turn); prev = held
    return pd.Series(rets, index=months), turns

def summarise(r, turns):
    ex = r - r.index.map(rf_series)
    return {"Annualised return (%)": 100 * ((1 + r / 100).prod() ** (12 / len(r)) - 1), "Annualised volatility (%)": r.std() * np.sqrt(12),
            "Sharpe ratio": ex.mean() / ex.std() * np.sqrt(12), "Mean one-way turnover": float(np.mean(turns)), "Months": len(r)}

FLOORS = {"no floor": None, "floor 50th pct": 0.5, "floor 80th pct": 0.8}
def picks_for(score_col, floor_q):
    out = {}
    for y in BT_YEARS:
        year_pool = bt[bt["Formation_Year"] == y]
        pool = year_pool if floor_q is None else year_pool[year_pool["Environmental_Score"] >= np.quantile(year_pool["Environmental_Score"], floor_q)]
        out[y] = pool.nlargest(max(int(0.1 * len(pool)), 1), score_col)["Company_ID"].tolist()
    return out

series, metrics, names_held = {}, {}, {}
for model in MODELS:
    for floor, q in FLOORS.items():
        picks = picks_for(model, q)
        series[(model, floor)], turns = run_rule(picks, returns_c)
        metrics[(model, floor)] = summarise(series[(model, floor)], turns)
        names_held[(model, floor)] = float(np.mean([len(v) for v in picks.values()]))
bench_picks = {y: bt.loc[bt["Formation_Year"] == y, "Company_ID"].tolist() for y in BT_YEARS}
series["bench"], bench_turns = run_rule(bench_picks, returns_c, cost_bp=0)
metrics["bench"] = summarise(series["bench"], bench_turns)

main_model = "Linear regression (selected)"
main_table = pd.DataFrame({**{f"{main_model}, {f}": metrics[(main_model, f)] for f in FLOORS}, "Equal-weight benchmark (all firms)": metrics["bench"]}).T
main_table["Names held (average)"] = [names_held[(main_model, f)] for f in FLOORS] + [float(np.mean([len(v) for v in bench_picks.values()]))]
print(f"Formation years {BT_YEARS[0]} to {BT_YEARS[-1]}, held July to June (84 months), 20 bp per unit of one-way turnover, full panel including firms that left:")
display(main_table.round(3))
b = metrics["bench"]
print("Difference to the benchmark (percentage points of annual return / Sharpe points): " + "; ".join(
    f"{f}: {metrics[(main_model, f)]['Annualised return (%)'] - b['Annualised return (%)']:+.2f} / {metrics[(main_model, f)]['Sharpe ratio'] - b['Sharpe ratio']:+.3f}" for f in FLOORS))
print("Cost of the floor against the same rule without a floor: " + "; ".join(
    f"{f}: {metrics[(main_model, f)]['Annualised return (%)'] - metrics[(main_model, 'no floor')]['Annualised return (%)']:+.2f} points of return, "
    f"{metrics[(main_model, f)]['Sharpe ratio'] - metrics[(main_model, 'no floor')]['Sharpe ratio']:+.3f} Sharpe" for f in ["floor 50th pct", "floor 80th pct"]))

print("Cost drag per year (mean one-way turnover x 40 bp): " + "; ".join(f"{f}: {0.4 * metrics[(main_model, f)]['Mean one-way turnover']:.2f} points" for f in FLOORS))
rob_rows = {f"{m}, {f}": metrics[(m, f)] for m in MODELS if m != main_model for f in FLOORS}
print("\nLabelled robustness checks (same design, other forecasts):")
display(pd.DataFrame(rob_rows).T.round(3))

late = lambda s: s[s.index >= "2022-07"]
late_rows = {f"{main_model}, {f}": summarise(late(series[(main_model, f)]), [np.nan]) for f in FLOORS}
late_rows["Equal-weight benchmark (all firms)"] = summarise(late(series["bench"]), [np.nan])
print("\nOnly the clean out-of-sample years (formation 2022 to 2024, July 2022 to June 2025, 36 months; turnover not shown):")
display(pd.DataFrame(late_rows).T.drop(columns="Mean one-way turnover").round(3))
```

```python
# ---- (d) Step 3: bootstrap of the Sharpe difference (months resampled, same months for rule and benchmark) ----
assert "series" in globals(), "Run the cells above first (Runtime > Run all)."
B, rng = 5000, np.random.default_rng(12345)
ex_bench = (series["bench"] - series["bench"].index.map(rf_series)).to_numpy()
n_months = len(ex_bench)
idx = rng.integers(0, n_months, size=(B, n_months))              # the same resampled months for every rule
sharpe_of = lambda ex: ex.mean(axis=-1) / ex.std(axis=-1, ddof=1) * np.sqrt(12)
boot = {}
for f in FLOORS:
    ex_rule = (series[(main_model, f)] - series[(main_model, f)].index.map(rf_series)).to_numpy()
    assert len(ex_rule) == n_months
    diffs = sharpe_of(ex_rule[idx]) - sharpe_of(ex_bench[idx])
    point = float(sharpe_of(ex_rule) - sharpe_of(ex_bench))
    lo, hi = np.percentile(diffs, [2.5, 97.5])
    boot[f] = dict(point=point, lo=float(lo), hi=float(hi), share_positive=float((diffs > 0).mean()))
boot_table = pd.DataFrame(boot).T
boot_table.index = [f"{main_model}, {f} minus benchmark" for f in boot_table.index]
print("Sharpe ratio difference (rule minus equal-weight benchmark), 95% percentile interval over 5,000 resamples of the 84 months:")
display(boot_table.round(3))
best_rule = max(FLOORS, key=lambda f: metrics[(main_model, f)]["Sharpe ratio"])
print(f"Best rule by Sharpe ratio: {main_model}, {best_rule}: difference {boot[best_rule]['point']:+.3f}, 95% interval {boot[best_rule]['lo']:+.3f} to {boot[best_rule]['hi']:+.3f}. "
      + ("The interval contains zero: the difference to the benchmark is not distinguishable from noise." if boot[best_rule]["lo"] <= 0 <= boot[best_rule]["hi"]
         else "The interval excludes zero."))
print("Caveats: choosing the best of three rules after seeing the results flatters it; resampling months independently ignores autocorrelation; 84 months are few.")
```

```python
# ---- (d) Step 4: figure: cumulative return series ----
assert "series" in globals(), "Run the cells above first (Runtime > Run all)."
cum = lambda s: (1 + s / 100).cumprod()
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5), sharey=True)
for f, colour in zip(FLOORS, ["steelblue", "darkorange", "seagreen"]):
    axes[0].plot(cum(series[(main_model, f)]).values, label=f"{main_model}, {f}", color=colour)
axes[0].plot(cum(series["bench"]).values, color="black", ls="--", label="Equal-weight benchmark (all firms)")
axes[0].set_title("Selected model (linear regression) with and without the ESG floor")
for (m, colour) in zip([m for m in MODELS if m != main_model], ["purple", "firebrick"]):
    axes[1].plot(cum(series[(m, "no floor")]).values, label=f"{m}, no floor", color=colour)
axes[1].plot(cum(series[(main_model, "no floor")]).values, label=f"{main_model}, no floor", color="steelblue")
axes[1].plot(cum(series["bench"]).values, color="black", ls="--", label="Equal-weight benchmark (all firms)")
axes[1].set_title("Labelled robustness checks (no floor)")
months = list(series["bench"].index)
for ax in axes:
    ax.set_xticks(range(0, len(months), 12)); ax.set_xticklabels([months[i] for i in range(0, len(months), 12)], rotation=45)
    ax.axvline(months.index("2022-07"), color="grey", ls=":", lw=1)
    ax.set_xlabel("Month (July 2018 to June 2025; dotted line: start of the 2022 to 2024 formation years)"); ax.legend(fontsize=8, loc="upper left")
axes[0].set_ylabel("Cumulative value of 1 invested in July 2018 (after costs)")
plt.tight_layout(); plt.show()
```

```python
# ---- (d) Step 5: the numbers for results.json ----
assert "boot" in globals(), "Run the cells above first (Runtime > Run all)."
def rule_row(name, m):
    return dict(name=name, ann_return_pct=float(m["Annualised return (%)"]), ann_vol_pct=float(m["Annualised volatility (%)"]),
                sharpe=float(m["Sharpe ratio"]), turnover=float(m["Mean one-way turnover"]))
task_d = dict(
    rules=[rule_row(f"{main_model}, {f}", metrics[(main_model, f)]) for f in FLOORS] + [rule_row("Equal-weight benchmark (all firms)", metrics["bench"])],
    bootstrap=dict(statistic=f"Sharpe ratio difference, {main_model}, {best_rule} minus equal-weight benchmark; 84 months resampled with replacement, 5,000 draws, seed 12345, 95% percentile interval",
                   point=boot[best_rule]["point"], lo=boot[best_rule]["lo"], hi=boot[best_rule]["hi"]),
    universe="full panel including firms that left")
for r_ in task_d["rules"]:
    print(r_)
print(task_d["bootstrap"])
```

**Reading the result of task (d)** *(numbers from the run on 3 November; re-check them if the notebook changes).*

*Does the selected rule beat the benchmark?* Not in a way that can be told apart from chance. Linear regression, no floor, earned 9.4% a year after costs against 8.8% for the equal-weight benchmark (+0.6 points; Sharpe ratio 0.46 against 0.42). The forecast itself has almost no ranking skill: the rank correlation with the realised return averages +0.01 over 2018 to 2024 (+0.03 in 2018 to 2021, -0.02 in 2022 to 2024). The strategy holds about 94 names against about 945 for the benchmark, so a difference of this size is within what a different, equally unskilled selection of names would produce. In the clean out-of-sample years (formation 2022 to 2024, a strong market) the picture is the same: 21.9% against 21.4%.

*What the floor does.* In this sample the floor did **not** cost return: the 50th-percentile floor earned 9.6% (+0.2 points against no floor), the 80th-percentile floor 11.4% (+2.0 points; Sharpe 0.53). What it costs is concentration and trading: about 47 and 19 names instead of 94, volatility of 17.1% and 18.4% instead of 16.8%, and one-way turnover of 0.76 and 0.82 instead of 0.70, that is a cost drag of 0.31 and 0.33 points a year (0.28 without a floor). The 80th-percentile result rests on about 19 names a year and should not be read as an ESG effect: the selected model has no ranking skill, and in task (c) no region-period cell other than Europe from 2021 shows a relation between the Environmental score and returns.

*Is the difference to the benchmark distinguishable from noise?* No. For the best rule (80th-percentile floor) the Sharpe difference is +0.11 with a 95% interval from -0.16 to +0.38; the intervals of the other two rules also contain zero (-0.16 to +0.23 and -0.18 to +0.26). Choosing the best of three rules after seeing the results flatters it, months are resampled independently although returns can be autocorrelated, and 84 months are few.

*Robustness checks (labelled, not chosen on cross-validation).* Gradient boosting without a floor earned 8.1% (Sharpe 0.39), below the benchmark. The linear variant with past return standardised within year earned 11.5% (Sharpe 0.57), the highest no-floor result, but it was not the model selected in task (b), and picking it now because it looks best would be selecting on the test period.

```python
import json

RESULTS_TEMPLATE = {
    "group": "",                    # e.g. "Group 7"
    "members": [],                  # full names of every group member
    "audit": [                      # one entry per problem you found and treated, in your own words
        {
            "issue": "",             # short description of what you found
            "file": "",               # which file it is in
            "detected_by": "",         # how you found it
            "treatment": "",           # what you did about it
            "affected_ids": [],        # Company_IDs it affects, if there is a natural list; may stay empty
            "affected_count": 0,       # how many rows or firms it affects
            "effect": "",               # which number it changed, and by how much
        },
    ],
    "task_b": {
        "target": "",                # what you forecast, e.g. "twelve-month holding return"
        "test_formation_years": [],  # formation years held out for the test set
        "baseline_rmse": 0.0,        # naive mean baseline, test RMSE
        "linear_rmse": 0.0,          # linear regression, test RMSE
        "ensemble_rmse": 0.0,        # your selected ensemble, test RMSE
        "ensemble_kind": "",         # "random forest" or "gradient boosting"
        "selected_on_cv": "",        # which model/setting cross-validation selected
        "test_r2": 0.0,              # selected model's test R2
    },
    "task_c": {
        "estimates": [               # point-in-time scores, one row per region x period
            {"region": "", "period": "", "coef": 0.0, "se": 0.0, "n": 0},
        ],
        "restated": [                # same cells, restated scores
            {"region": "", "period": "", "coef": 0.0, "se": 0.0, "n": 0},
        ],
        "conclusion": "",            # where an investor could have earned a premium, and where not
    },
    "task_d": {
        "rules": [                   # one row per rule: no floor, each floor, benchmark
            {"name": "", "ann_return_pct": 0.0, "ann_vol_pct": 0.0, "sharpe": 0.0, "turnover": 0.0},
        ],
        "bootstrap": {"statistic": "", "point": 0.0, "lo": 0.0, "hi": 0.0},  # what you bootstrapped, and the interval
        "universe": "",              # "full panel including firms that left" or "survivors only"
    },
}


def _check(data, template, path):
    # Walks RESULTS_TEMPLATE's shape and checks data against it: every key in the template must be
    # present in data, and every numeric leaf in the template must be a number (not a string) in data.
    if isinstance(template, dict):
        if not isinstance(data, dict):
            raise ValueError(f"results.json: '{path}' should be an object, got {type(data).__name__}")
        for key, sub_template in template.items():
            sub_path = f"{path}.{key}" if path else key
            if key not in data:
                raise ValueError(f"results.json: missing required key '{sub_path}'")
            _check(data[key], sub_template, sub_path)
    elif isinstance(template, list):
        if not isinstance(data, list):
            raise ValueError(f"results.json: '{path}' should be a list, got {type(data).__name__}")
        if template and data:
            item_template = template[0]
            for i, item in enumerate(data):
                _check(item, item_template, f"{path}[{i}]")
    elif isinstance(template, (int, float)) and not isinstance(template, bool):
        if isinstance(data, bool) or not isinstance(data, (int, float)):
            raise ValueError(f"results.json: '{path}' should be a number, got {type(data).__name__} ({data!r})")


def write_results(data, path="results.json"):
    # Validates data against the submission contract's schema, writes it, and prints what it wrote.
    # Raises ValueError naming the exact key that is missing or has the wrong type.
    _check(data, RESULTS_TEMPLATE, "")
    with open(path, "w") as f:
        json.dump(data, f, indent=2)
    print(f"wrote {path}")
    print(f"  group: {data['group'] or '(not set)'}, {len(data['members'])} member(s)")
    print(f"  audit: {len(data['audit'])} entr{'y' if len(data['audit']) == 1 else 'ies'}")
    print(f"  task_b: selected on CV = {data['task_b']['selected_on_cv'] or '(not set)'}, test R2 = {data['task_b']['test_r2']}")
    print(f"  task_c: {len(data['task_c']['estimates'])} point-in-time estimate(s), {len(data['task_c']['restated'])} restated")
    print(f"  task_d: {len(data['task_d']['rules'])} rule(s), universe = {data['task_d']['universe'] or '(not set)'}")


# Run once now, on the template as it ships, so you see a valid file before you change a single value.
# Fill in RESULTS_TEMPLATE (or build your own dict with the same keys) with your own numbers, then call
# write_results again as the last cell of your notebook.
write_results(RESULTS_TEMPLATE)
```
